```python
import pandas as pd
import numpy as np
from IPython.display import display, HTML

# Diễn giải thuật toán:
# 1. Đo lường tỷ lệ trùng lặp CALL_ID để đảm bảo granularity (độ hạt) là 1 dòng = 1 sự kiện gọi.
# 2. Left Anti Join (hoặc set difference) để đếm số lượng SKP_CLIENT có trong CALL nhưng không có trong gốc CUSTOMER.

def check_keys_integrity(df_call, df_customer):
    # 1. Kiểm tra duy nhất CALL_ID
    total_calls = len(df_call)
    unique_calls = df_call['CALL_ID'].nunique()
    is_call_unique = total_calls == unique_calls

    # 2. Tỷ lệ khớp nối SKP_CLIENT
    call_clients = set(df_call['SKP_CLIENT'].unique())
    base_clients = set(df_customer['SKP_CLIENT'].unique())

    missing_clients = call_clients - base_clients
    match_rate = (len(call_clients) - len(missing_clients)) / len(call_clients) * 100

    # Thiết kế bảng HTML phân nhóm logic chặt chẽ theo Corporate Theme
    # Đảm bảo không để trống bất kỳ giá trị nào
    status_unique = '✓ ĐẠT (Duy nhất)' if is_call_unique else '⚠ KHÔNG ĐẠT (Trùng lặp)'
    status_color = '#2E7D32' if is_call_unique else '#C62828'
    missing_color = '#2E7D32' if len(missing_clients) == 0 else '#C62828'

    html_output = f"""
    <div style="font-family: Arial, sans-serif; max-width: 650px; margin: 15px 0; border: 1px solid #17365D; border-radius: 8px; overflow: hidden; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">

        <!-- Header chính -->
        <div style="background-color: #17365D; color: white; padding: 14px 18px; font-weight: bold; font-size: 16px; letter-spacing: 0.5px;">
            KIỂM TOÁN TÍNH TOÀN VẸN VÀ ĐỘ HẠT CỦA KHÓA CHÍNH TRONG CALL.csv
        </div>

        <table style="width: 100%; border-collapse: collapse; font-size: 14px; text-align: left; background-color: #ffffff;">

            <!-- PHẦN 1: ĐỘ HẠT CUỘC GỌI -->
            <tr style="background-color: #EAF2F8; border-bottom: 2px solid #17365D;">
                <th colspan="2" style="padding: 10px 16px; color: #17365D; font-weight: bold; font-size: 14px; text-transform: uppercase;">
                    1. Tính duy nhất của sự kiện gọi (Granularity)
                </th>
            </tr>
            <tr style="border-bottom: 1px solid #D9E2F3;">
                <td style="padding: 10px 16px; width: 60%; color: #333;">Tổng số bản ghi cuộc gọi (CALL_ID)</td>
                <td style="padding: 10px 16px; text-align: right; font-family: monospace; font-weight: bold;">{total_calls:,}</td>
            </tr>
            <tr style="border-bottom: 1px solid #D9E2F3;">
                <td style="padding: 10px 16px; color: #333;">Số cuộc gọi thực tế duy nhất (Unique CALL_ID)</td>
                <td style="padding: 10px 16px; text-align: right; font-family: monospace; font-weight: bold;">{unique_calls:,}</td>
            </tr>
            <tr style="border-bottom: 2px solid #17365D;">
                <td style="padding: 10px 16px; color: #333;">Trạng thái đảm bảo khóa chính (No Duplicates)</td>
                <td style="padding: 10px 16px; text-align: right; font-weight: bold; color: {status_color};">
                    {status_unique}
                </td>
            </tr>

            <!-- PHẦN 2: LIÊN KẾT KHÁCH HÀNG -->
            <tr style="background-color: #EAF2F8; border-bottom: 2px solid #17365D;">
                <th colspan="2" style="padding: 10px 16px; color: #17365D; font-weight: bold; font-size: 14px; text-transform: uppercase;">
                    2. Tính toàn vẹn của liên kết dữ liệu (Data Integrity)
                </th>
            </tr>
            <tr style="border-bottom: 1px solid #D9E2F3;">
                <td style="padding: 10px 16px; color: #333;">Số lượng Khách hàng duy nhất phát sinh cuộc gọi</td>
                <td style="padding: 10px 16px; text-align: right; font-family: monospace; font-weight: bold;">{len(call_clients):,}</td>
            </tr>
            <tr style="border-bottom: 1px solid #D9E2F3;">
                <td style="padding: 10px 16px; color: #333;">Số lượng Khách hàng mồ côi (Không khớp CUSTOMER)</td>
                <td style="padding: 10px 16px; text-align: right; font-family: monospace; font-weight: bold; color: {missing_color};">
                    {len(missing_clients):,}
                </td>
            </tr>
            <tr style="background-color: #F2F5F9;">
                <td style="padding: 12px 16px; font-weight: bold; color: #17365D; font-size: 15px;">Tỷ lệ khớp nối thành công (Match Rate)</td>
                <td style="padding: 12px 16px; text-align: right; font-weight: bold; color: {'#2E7D32' if match_rate == 100 else '#17365D'}; font-size: 16px;">
                    {match_rate:.2f}%
                </td>
            </tr>
        </table>
    </div>
    """
    display(HTML(html_output))
    return missing_clients

# Thực thi
missing_clients_call = check_keys_integrity(df_call, df_customer)
```

```python
# XEM PHÂN PHỐI CỦA DURATION

# ==============================================================================
# CÀI ĐẶT THEME DESIGN
# ==============================================================================
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['text.color'] = '#17365D'
plt.rcParams['axes.labelcolor'] = '#17365D'
plt.rcParams['xtick.color'] = '#17365D'
plt.rcParams['ytick.color'] = '#17365D'

COLOR_PRIMARY = '#17365D'
COLOR_NEUTRAL = '#B4C7E7'
COLOR_ALERT   = '#e74c3c'
COLOR_GRID    = '#D9E2F3'
COLOR_BASELINE= '#34495e'

# ==============================================================================
# 1. TÍNH TOÁN THỐNG KÊ MÔ TẢ CHO DURATIONS
# ==============================================================================
# Giả định df_call đã được nạp từ CALL.csv
durations = df_call['DURATIONS'].dropna()

stats_dict = {
    'Tổng số bản ghi (Count)': len(durations),
    'Trung bình (Mean)': durations.mean(),
    'Độ lệch chuẩn (Std)': durations.std(),
    'Nhỏ nhất (Min)': durations.min(),
    'Phân vị 25% (Q1)': durations.quantile(0.25),
    'Trung vị 50% (Median)': durations.median(),
    'Phân vị 75% (Q3)': durations.quantile(0.75),
    'Lớn nhất (Max)': durations.max(),
    'Số lượng DURATIONS < 0 (Lỗi vật lý)': (durations < 0).sum(),
    'Số lượng DURATIONS = 0 (Không đàm thoại)': (durations == 0).sum()
}

df_stats = pd.DataFrame(list(stats_dict.items()), columns=['Chỉ tiêu Thống kê', 'Giá trị'])

# Định dạng số liệu
def format_val(x):
    if isinstance(x, int) or x.is_integer():
        return f"{int(x):,}"
    return f"{x:,.2f}"

df_stats['Giá trị'] = df_stats['Giá trị'].apply(format_val)

# ==============================================================================
# 2. XUẤT BẢNG HTML CHUẨN THEME
# ==============================================================================
html_table = f"""
<div style="font-family: Arial, sans-serif; color: {COLOR_PRIMARY}; width: 100%; margin-bottom: 20px;">
    <h3 style="color: {COLOR_PRIMARY}; font-size: 16px; font-weight: bold; text-align: left; padding-bottom: 10px; margin: 0;">
        BẢNG THỐNG KÊ MÔ TẢ BIẾN DURATIONS (THỜI LƯỢNG ĐÀM THOẠI THỰC TẾ)
    </h3>
    <table style="width: 100%; border-collapse: collapse; border: 1px solid {COLOR_GRID}; font-size: 13px;">
        <thead>
            <tr style="background-color: {COLOR_PRIMARY}; color: #FFFFFF;">
                <th style="padding: 10px 14px; text-align: left; border: 1px solid {COLOR_GRID};">Chỉ tiêu Thống kê</th>
                <th style="padding: 10px 14px; text-align: right; border: 1px solid {COLOR_GRID};">Giá trị (Giây)</th>
            </tr>
        </thead>
        <tbody>
"""
for idx, row in df_stats.iterrows():
    # Highlight dòng lỗi hoặc không đàm thoại
    if "Lỗi" in row['Chỉ tiêu Thống kê'] or "=" in row['Chỉ tiêu Thống kê']:
        bg_color = "#D9EAF7"
        font_weight = "bold"
        text_color = COLOR_ALERT if "Lỗi" in row['Chỉ tiêu Thống kê'] else COLOR_PRIMARY
    else:
        bg_color = "#EAF2F8" if idx % 2 == 0 else "#FFFFFF"
        font_weight = "normal"
        text_color = COLOR_PRIMARY

    html_table += f"""
            <tr style="background-color: {bg_color}; color: {text_color};">
                <td style="padding: 9px 14px; text-align: left; border: 1px solid {COLOR_GRID}; font-weight: bold;">{row['Chỉ tiêu Thống kê']}</td>
                <td style="padding: 9px 14px; text-align: right; border: 1px solid {COLOR_GRID}; font-weight: {font_weight};">{row['Giá trị']}</td>
            </tr>
    """
html_table += "</tbody></table></div>"
display(HTML(html_table))

# ==============================================================================
# 3. VẼ BIỂU ĐỒ PHÂN PHỐI & OUTLIERS (WIDE CANVAS)[cite: 6]
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))

# Loại bỏ giá trị âm để vẽ biểu đồ phân phối chuẩn
valid_durations = durations[durations >= 0]

# Biểu đồ 1: Histogram
sns.histplot(valid_durations, bins=50, kde=True, color=COLOR_NEUTRAL, edgecolor='white', ax=axes[0])
axes[0].axvline(valid_durations.mean(), color=COLOR_ALERT, linestyle='--', linewidth=1.5, label=f'Mean: {valid_durations.mean():.1f}s')
axes[0].axvline(valid_durations.median(), color=COLOR_PRIMARY, linestyle='-', linewidth=1.5, label=f'Median: {valid_durations.median():.1f}s')
axes[0].set_title('PHÂN PHỐI THỜI LƯỢNG ĐÀM THOẠI (DURATIONS >= 0)', loc='left', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Thời lượng đàm thoại (Giây)', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Số lượng cuộc gọi', fontsize=11, fontweight='bold')
axes[0].axhline(0, color=COLOR_BASELINE, linestyle='--', linewidth=1.2, alpha=0.7)
axes[0].legend(loc='upper right')
axes[0].grid(axis='y', color=COLOR_GRID, linestyle='-', alpha=0.3)

# Biểu đồ 2: Boxplot
sns.boxplot(x=valid_durations, color=COLOR_NEUTRAL, flierprops=dict(markerfacecolor=COLOR_ALERT, marker='o', markersize=3), ax=axes[1])
axes[1].set_title('NHẬN DIỆN ĐIỂM DỊ BIỆT (BOXPLOT)', loc='left', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Thời lượng đàm thoại (Giây)', fontsize=11, fontweight='bold')
axes[1].grid(axis='x', color=COLOR_GRID, linestyle='-', alpha=0.3)

# Lược bỏ khung viền
sns.despine(ax=axes[0])
sns.despine(ax=axes[1])

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from html import escape
from textwrap import fill
from matplotlib import font_manager
from IPython.display import HTML, display

# ============================================================
# UNIFIED BI THEME
# ============================================================
NAVY = "#17365D"
BLUE = "#85C1E9"
NEUTRAL = "#B4C7E7"
GRID = "#D9E2F3"

font_name = (
    "Arial"
    if any(f.name == "Arial" for f in font_manager.fontManager.ttflist)
    else "DejaVu Sans"
)

plt.rcParams.update({
    "font.family": font_name,
    "text.color": NAVY,
    "axes.labelcolor": NAVY,
    "xtick.color": NAVY,
    "ytick.color": NAVY,
    "font.size": 11,
})

# Bảng HTML chuẩn theme, có title và methodological notes.
def bi_table(data, title, note="", formats=None):
    styled = (
        data.style
        .format(formats or {}, na_rep="—")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("width", "100%"),
                    ("border-collapse", "collapse"),
                    ("font-family", "Arial, sans-serif"),
                    ("font-size", "13px"),
                ],
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", NAVY),
                    ("color", "#FFFFFF"),
                    ("padding", "10px 14px"),
                    ("text-align", "left"),
                ],
            },
            {
                "selector": "td",
                "props": [
                    ("color", NAVY),
                    ("padding", "9px 14px"),
                    ("border-bottom", f"1px solid {GRID}"),
                    ("text-align", "right"),
                ],
            },
            {
                "selector": "td:first-child",
                "props": [("text-align", "left")],
            },
        ])
        .hide(axis="index")
    )

    note_html = escape(note).replace("\n", "<br>")

    display(HTML(
        f'<div style="font-family:Arial,sans-serif;color:{NAVY};">'
        f'<h3 style="font-size:17px;margin:0 0 12px 0;">'
        f'{escape(title)}</h3>'
        f'{styled.to_html()}'
        f'<p style="font-size:12px;line-height:1.5;">'
        f'{note_html}</p></div>'
    ))


# Biểu đồ cột ngang có sẵn Data Labels, mẫu số và chú thích.
def bi_bar(values, title, note, denominator=None):
    values = values.astype("int64")
    fig, ax = plt.subplots(
        figsize=(12, max(4.4, 0.55 * len(values) + 2.5))
    )

    bars = ax.barh(
        values.index.astype(str),
        values.to_numpy(),
        color=[NAVY] + [NEUTRAL] * max(len(values) - 1, 0),
        height=0.58,
        zorder=3,
    )

    labels = [
        f"{v:,} ({v / denominator:.2%})"
        if denominator else f"{v:,}"
        for v in values
    ]

    ax.bar_label(
        bars,
        labels=labels,
        padding=6,
        fontsize=11,
        fontweight="bold",
        color=NAVY,
    )

    ax.invert_yaxis()
    ax.set_xlim(0, max(int(values.max()), 1) * 1.40)
    ax.set_xlabel("Number of Records", fontweight="bold")
    ax.grid(axis="x", color=GRID, alpha=0.3, zorder=0)
    ax.set_axisbelow(True)
    ax.spines[["top", "right"]].set_visible(False)

    fig.suptitle(
        title,
        x=0.04,
        y=0.97,
        ha="left",
        fontsize=16,
        fontweight="bold",
        color=NAVY,
    )

    note_text = "\n".join(
        fill(line, width=115) for line in note.split("\n")
    )

    fig.text(
        0.04, 0.045,
        note_text,
        ha="left",
        va="bottom",
        fontsize=10,
        color=NAVY,
        linespacing=1.5,
    )

    fig.subplots_adjust(
        left=0.34, right=0.95, top=0.82, bottom=0.25
    )
    plt.show()
```

```python

# ============================================================
# STEP 1 — TIMESTAMP VALIDATION
# ============================================================
GROUP_KEYS = ["SKP_CLIENT", "ACTION", "PRODUCT_CODE"]
REQUIRED = GROUP_KEYS + ["DTIME_EVENT"]

missing = set(REQUIRED) - set(df_past_behavior.columns)
if missing:
    raise KeyError(f"Missing columns: {sorted(missing)}")

# Tạo DataFrame riêng, bảo toàn toàn bộ cột dữ liệu nguồn.
df_events = (
    df_past_behavior
    .assign(
        DTIME_EVENT_PARSED=pd.to_datetime(
            df_past_behavior["DTIME_EVENT"],
            errors="coerce",
            utc=True,
        )
    )
    .sort_values(
        ["SKP_CLIENT", "DTIME_EVENT_PARSED"],
        kind="stable",
        na_position="last",
        ignore_index=True,
    )
)

raw_count = len(df_past_behavior)
valid_time = int(df_events["DTIME_EVENT_PARSED"].notna().sum())

# Sanity check
assert len(df_events) == raw_count
assert valid_time + df_events["DTIME_EVENT_PARSED"].isna().sum() == raw_count

step1_summary = pd.DataFrame({
    "Timestamp Quality Metric": [
        "Raw Events",
        "Valid Timestamps",
        "Missing / Unparseable Timestamps",
    ],
    "Count": [
        raw_count,
        valid_time,
        raw_count - valid_time,
    ],
})

step1_summary["Share (%)"] = (
    step1_summary["Count"] / max(raw_count, 1) * 100
)

bi_table(
    step1_summary,
    "STEP 1 | EVENT TIMESTAMP VALIDATION",
    note=(
        "Definition: Valid Timestamp = successfully parsed DTIME_EVENT.\n"
        "Denominator: All raw events. No event is removed.\n"
        "Timezone convention: UTC; source timezone must be verified."
    ),
    formats={"Count": "{:,.0f}", "Share (%)": "{:.2f}%"},
)
```

```python

# ============================================================
# STEP 2 — OBSERVED ACTION CLASSIFICATION
# ============================================================
action_summary = (
    df_events["ACTION"]
    .value_counts(dropna=False)
    .rename_axis("ACTION")
    .reset_index(name="Event_Count")
)

action_summary["Share (%)"] = (
    action_summary["Event_Count"] / max(raw_count, 1) * 100
)

# Nhãn hiển thị; không thay đổi ACTION trong df_events.
action_summary["ACTION_LABEL"] = (
    action_summary["ACTION"]
    .astype("string")
    .fillna("[NULL]")
)

# Sanity check: mọi ACTION, kể cả missing, đều được bảo toàn.
assert action_summary["Event_Count"].sum() == raw_count

# Biểu đồ tối đa 10 nhãn để giữ khả năng đọc trên slide.
top_actions = (
    action_summary.head(10)
    .set_index("ACTION_LABEL")["Event_Count"]
)

if not top_actions.empty:
    bi_bar(
        top_actions,
        "STEP 2 | OBSERVED ACTION DISTRIBUTION",
        note=(
            "Action Share (%) = Action Event Count / All Raw Events × 100.\n"
            "Classification: Exact observed ACTION labels only; "
            "[NULL] denotes missing values. No semantic mapping applied."
            + (
                "\nChart shows the 10 most frequent labels."
                if len(action_summary) > 10 else ""
            )
        ),
        denominator=raw_count,
    )

# Hiển thị đầy đủ nếu danh mục vượt giới hạn biểu đồ.
if len(action_summary) > 10:
    bi_table(
        action_summary[["ACTION_LABEL", "Event_Count", "Share (%)"]],
        "FULL ACTION VALUE CATALOG",
        note="Source: Observed ACTION values, including missing.",
        formats={
            "Event_Count": "{:,.0f}",
            "Share (%)": "{:.2f}%",
        },
    )
```

```python

# ============================================================
# STEP 3 — INTER-EVENT GAP ANALYSIS
# ============================================================
SUSPECT_GAP_SEC = 60
SESSION_GAP_SEC = 300

# Lấy timestamp trước đó trong cùng nhóm, không so nhóm khác.
previous_time = (
    df_events.groupby(
        GROUP_KEYS, sort=False, dropna=False
    )["DTIME_EVENT_PARSED"]
    .shift()
)

# Không tính gap nếu khách hàng / ACTION / sản phẩm bị thiếu.
invalid_key = df_events[GROUP_KEYS].isna().any(axis=1)

df_events["EVENT_GAP_SEC"] = (
    df_events["DTIME_EVENT_PARSED"]
    .sub(previous_time)
    .dt.total_seconds()
    .mask(invalid_key)
)

# Chỉ đánh dấu, tuyệt đối không xóa log.
df_events["SUSPECT_GAP_FLAG"] = (
    df_events["EVENT_GAP_SEC"].le(SUSPECT_GAP_SEC)
)

gap = df_events["EVENT_GAP_SEC"]
valid_pairs = int(gap.notna().sum())

gap_counts = pd.Series({
    "Gap = 0s": int(gap.eq(0).sum()),
    "Gap <= 60s": int(gap.le(60).sum()),
    "Gap <= 300s": int(gap.le(300).sum()),
})

# Sanity check
assert gap.dropna().ge(0).all()
assert gap_counts.is_monotonic_increasing
assert len(df_events) == raw_count

bi_bar(
    gap_counts,
    "STEP 3 | INTER-EVENT GAP ANALYSIS",
    note=(
        "Formula: Gap_i (seconds) = Event Time_i - Previous Event Time "
        "within the same SKP_CLIENT + ACTION + PRODUCT_CODE.\n"
        f"Denominator: {valid_pairs:,} comparable event pairs. "
        "Threshold counts are cumulative, not mutually exclusive.\n"
        "Gap <= 60s flags temporal proximity only; "
        "it does not establish duplicate or system-error status."
    ),
    denominator=valid_pairs,
)
```

```python

# ============================================================
# STEP 4 — ROLLING INACTIVITY SESSIONIZATION
# ============================================================

# Mở phiên mới nếu không có gap hợp lệ hoặc gap > 300s.
# Log thiếu thời gian / khóa được giữ thành phiên đơn.
new_session = (
    gap.isna() | gap.gt(SESSION_GAP_SEC)
)

# Mỗi phiên nhận ID từ vị trí event đầu tiên của phiên.
session_anchor = pd.Series(
    np.arange(1, raw_count + 1, dtype="int64"),
    index=df_events.index,
).where(new_session)

# Forward-fill trong cùng bộ khóa, không gộp khác sản phẩm.
df_events["SESSION_ID"] = (
    session_anchor.groupby(
        [df_events[col] for col in GROUP_KEYS],
        sort=False,
        dropna=False,
    )
    .ffill()
    .astype("int64")
)

# Một dòng đại diện một phiên tương tác.
df_sessions = (
    df_events.groupby("SESSION_ID", sort=False)
    .agg(
        SKP_CLIENT=("SKP_CLIENT", "first"),
        ACTION=("ACTION", "first"),
        PRODUCT_CODE=("PRODUCT_CODE", "first"),
        Session_Start=("DTIME_EVENT_PARSED", "min"),
        Session_End=("DTIME_EVENT_PARSED", "max"),
        Event_Count=("SESSION_ID", "size"),
        Suspect_Event_Count=("SUSPECT_GAP_FLAG", "sum"),
    )
    .reset_index()
)

# ============================================================
# SANITY CHECK
# ============================================================
total_sessions = len(df_sessions)

# Mọi event được gán đúng một session.
assert df_events["SESSION_ID"].notna().all()
assert df_sessions["Event_Count"].sum() == raw_count
assert total_sessions == int(new_session.sum())

# Mỗi session chỉ có một khách hàng, ACTION và PRODUCT_CODE.
assert (
    df_events.groupby("SESSION_ID")[GROUP_KEYS]
    .nunique(dropna=False)
    .le(1)
    .all()
    .all()
)

step4_summary = pd.DataFrame({
    "Sessionization Metric": [
        "Raw Events",
        "Total Sessions",
        "Multi-event Sessions",
        "Events Preserved",
    ],
    "Count": [
        raw_count,
        total_sessions,
        int(df_sessions["Event_Count"].gt(1).sum()),
        int(df_sessions["Event_Count"].sum()),
    ],
})

bi_table(
    step4_summary,
    "STEP 4 | ROLLING SESSIONIZATION VALIDATION",
    note=(
        "Session Key = SKP_CLIENT + ACTION + PRODUCT_CODE.\n"
        "New Session = first valid event, missing/uncomparable gap, "
        "or rolling inactivity gap > 300 seconds.\n"
        "Consecutive gap <= 300 seconds preserves the current session. "
        "Missing time/key events form singleton sessions.\n"
        "Conservation: Sum(Session Event_Count) = Raw Events. "
        "No event is deleted."
    ),
    formats={"Count": "{:,.0f}"},
)
```

```python

# ============================================================
# STEP 5 — CUSTOMER-LEVEL FEATURES
# ============================================================

# Snapshot dựa trên thời điểm quan sát hợp lệ mới nhất.
snapshot_time = df_events["DTIME_EVENT_PARSED"].max()

if pd.isna(snapshot_time):
    raise ValueError("Cannot calculate Recency: no valid timestamp.")

snapshot_date = snapshot_time.normalize()

# Customer-level: một dòng trên một SKP_CLIENT hợp lệ.
df_customer_features = (
    df_events.groupby("SKP_CLIENT", sort=False)
    .agg(
        Session_Frequency=("SESSION_ID", "nunique"),
        Last_Event=("DTIME_EVENT_PARSED", "max"),
        Unique_ACTION_Count=("ACTION", "nunique"),
        Unique_PRODUCT_CODE_Count=("PRODUCT_CODE", "nunique"),
    )
    .reset_index()
)

# Recency = Snapshot Date - Latest Event Date.
# Không có thời gian hợp lệ => <NA>, không tự gán bằng 0.
df_customer_features["Recency_Days"] = (
    snapshot_date
    - df_customer_features["Last_Event"].dt.normalize()
).dt.days.astype("Int32")

# ============================================================
# FINAL VALIDATION
# ============================================================
raw_customers = df_past_behavior["SKP_CLIENT"].nunique()
output_customers = df_customer_features["SKP_CLIENT"].nunique()

compression_rate = (
    1 - total_sessions / max(raw_count, 1)
) * 100

# Khách có ít nhất một session chứa trên một event.
affected_customers = (
    df_sessions.loc[
        df_sessions["Event_Count"].gt(1),
        "SKP_CLIENT",
    ].nunique()
)

# Bảo toàn event và khách hàng có định danh.
assert df_sessions["Event_Count"].sum() == raw_count
assert raw_customers == output_customers

# Tổng session của khách bằng số session có SKP_CLIENT hợp lệ.
assert (
    df_customer_features["Session_Frequency"].sum()
    == df_sessions["SKP_CLIENT"].notna().sum()
)

assert df_customer_features["Recency_Days"].dropna().ge(0).all()

validation_summary = pd.DataFrame({
    "Validation Metric": [
        "Raw Events",
        "Total Sessions",
        "Compression Rate",
        "Affected Customers",
        "Raw Unique Customers",
        "Output Unique Customers",
        "Event Conservation",
        "Customer Conservation",
        "Snapshot Date (UTC)",
    ],
    "Value": [
        f"{raw_count:,}",
        f"{total_sessions:,}",
        f"{compression_rate:.2f}%",
        f"{affected_customers:,}",
        f"{raw_customers:,}",
        f"{output_customers:,}",
        "PASS",
        "PASS",
        str(snapshot_date.date()),
    ],
})

# ============================================================
# VISUAL OUTPUT — SLIDE-READY
# ============================================================
bi_bar(
    pd.Series({
        "Raw Events": raw_count,
        "Total Sessions": total_sessions,
    }),
    "PAST BEHAVIOR | EVENT-TO-SESSION COMPRESSION",
    note=(
        "Compression Rate (%) = "
        "(Raw Events - Total Sessions) / Raw Events × 100.\n"
        "Session: same SKP_CLIENT + ACTION + PRODUCT_CODE, "
        "rolling inactivity gap <= 300 seconds.\n"
        f"Affected Customers = {affected_customers:,} unique customers "
        "with at least one multi-event session. "
        "All raw events are preserved."
    ),
    denominator=raw_count,
)

# ============================================================
# HTML OUTPUT — VALIDATION SUMMARY
# ============================================================
bi_table(
    validation_summary,
    "FINAL VALIDATION | EVENT-TO-CUSTOMER PIPELINE",
    note=(
        "Compression measures reduction in record count, not deleted logs.\n"
        "Affected Customers = distinct SKP_CLIENT with >= 1 multi-event "
        "session. Missing SKP_CLIENT events remain at Event/Session level.\n"
        "Conservation excludes missing customer identifiers "
        "from the unique-customer comparison."
    ),
)

# ============================================================
# HTML OUTPUT — CUSTOMER FEATURE PREVIEW
# ============================================================
bi_table(
    df_customer_features[
        [
            "SKP_CLIENT",
            "Session_Frequency",
            "Recency_Days",
            "Unique_ACTION_Count",
            "Unique_PRODUCT_CODE_Count",
        ]
    ].head(5),
    "CUSTOMER ENGAGEMENT FEATURES | PREVIEW",
    note=(
        f"Snapshot Date (UTC): {snapshot_date.date()}.\n"
        "Session Frequency = Count(Distinct SESSION_ID).\n"
        "Recency (days) = Snapshot Date - Last Valid Event Date.\n"
        "Unique ACTION / PRODUCT_CODE = distinct non-null values "
        "per customer. Missing Recency remains unknown."
    ),
    formats={
        "Session_Frequency": "{:,.0f}",
        "Recency_Days": "{:,.0f}",
        "Unique_ACTION_Count": "{:,.0f}",
        "Unique_PRODUCT_CODE_Count": "{:,.0f}",
    },
)
```

```python

# ============================================================
# STEP 2 — ACTUAL LEAD SLA
# ============================================================
created = LEADS_CLEAN["DTIME_CREATED_VALID"]
accepted = LEADS_CLEAN["DATE_ACCEPTED_VALID"]

# Chỉ coi là time-precise khi CẢ HAI nguồn ghi rõ giờ/phút.
clock_pattern = r"(?:[ T])\d{1,2}:\d{2}"

has_clock = (
    df_lead["DTIME_CREATED"]
    .astype("string")
    .str.contains(clock_pattern, na=False)
    & df_lead["DATE_ACCEPTED"]
    .astype("string")
    .str.contains(clock_pattern, na=False)
)

# Đo bằng ngày thực khi đủ độ chính xác.
sla_exact = accepted.sub(created).dt.total_seconds().div(86400)

# Chỉ dùng ngày lịch khi ít nhất một đầu mốc thiếu giờ/phút.
sla_calendar = (
    accepted.dt.normalize()
    .sub(created.dt.normalize())
    .dt.days
)

LEADS_CLEAN["Actual_Lead_SLA"] = (
    sla_exact.where(has_clock, sla_calendar)
)

LEADS_CLEAN["SLA_Negative_Flag"] = (
    LEADS_CLEAN["Actual_Lead_SLA"].lt(0)
)

# Median / P90 chỉ trên SLA không âm, không missing.
valid_sla = LEADS_CLEAN.loc[
    LEADS_CLEAN["Actual_Lead_SLA"].ge(0),
    "Actual_Lead_SLA",
]

# ============================================================
# STEP 2 — SANITY CHECK
# ============================================================
assert LEADS_CLEAN.loc[
    LEADS_CLEAN["Is_Sentinel_3000_Flag"],
    "Actual_Lead_SLA",
].isna().all()

step2 = pd.DataFrame({
    "SLA Metric": [
        "Computable SLA",
        "SLA with Time Precision",
        "SLA with Calendar-Day Precision",
        "Negative SLA",
        "Missing SLA",
        "Median SLA (days)",
        "P90 SLA (days)",
    ],
    "Value": [
        f"{LEADS_CLEAN['Actual_Lead_SLA'].notna().sum():,}",
        f"{(LEADS_CLEAN['Actual_Lead_SLA'].notna() & has_clock).sum():,}",
        f"{(LEADS_CLEAN['Actual_Lead_SLA'].notna() & ~has_clock).sum():,}",
        f"{LEADS_CLEAN['SLA_Negative_Flag'].sum():,}",
        f"{LEADS_CLEAN['Actual_Lead_SLA'].isna().sum():,}",
        f"{valid_sla.median():,.2f}",
        f"{valid_sla.quantile(0.90):,.2f}",
    ],
})

bi_table(
    step2,
    "STEP 2 | ACTUAL LEAD SLA VALIDATION",
    note=(
        "Actual Lead SLA (days) = DATE_ACCEPTED_VALID - DTIME_CREATED_VALID.\n"
        "Timestamp precision: elapsed seconds / 86,400. "
        "Date-only precision: calendar-day difference.\n"
        "Median = P50; P90 = 90th percentile. Both exclude negative and "
        "missing SLA. Negative observations are flagged, not removed."
    ),
)
```

```python

# ============================================================
# STEP 3 — CREDIT FUNNEL VALIDATION
# ============================================================
apply_dt = CONTRACT_CLEAN["APPLY_CONTRACT_TIME_VALID"]
approve_dt = CONTRACT_CLEAN["APPROVE_CONTRACT_TIME_VALID"]
sign_dt = CONTRACT_CLEAN["SIGN_CONTRACT_TIME_VALID"]

# NaT không tạo vi phạm; chỉ so sánh các mốc có giá trị.
CONTRACT_CLEAN["Apply_Approve_Violation"] = apply_dt.gt(approve_dt)
CONTRACT_CLEAN["Approve_Sign_Violation"] = approve_dt.gt(sign_dt)
CONTRACT_CLEAN["Apply_Sign_Violation"] = apply_dt.gt(sign_dt)

CONTRACT_CLEAN["Funnel_Order_Violation"] = (
    CONTRACT_CLEAN["Apply_Approve_Violation"]
    | CONTRACT_CLEAN["Approve_Sign_Violation"]
    | CONTRACT_CLEAN["Apply_Sign_Violation"]
)

# Nhận diện signed status theo danh sách metadata.
signed_statuses = {
    "active", "finished", "paid off", "signed", "written off"
}
known_statuses = signed_statuses | {"others"}

status = (
    df_contract["NAME_CREDIT_STATUS"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

CONTRACT_CLEAN["Is_Signed_Status"] = (
    status.isin(signed_statuses)
    .where(status.isin(known_statuses), pd.NA)
    .astype("boolean")
)

# Trạng thái signed nhưng không có thời điểm ký hợp lệ,
# hoặc trạng thái Others nhưng lại có thời điểm ký.
CONTRACT_CLEAN["Signed_Time_Conflict_Flag"] = (
    (
        CONTRACT_CLEAN["Is_Signed_Status"]
        .eq(True).fillna(False)
        & sign_dt.isna()
    )
    |
    (
        CONTRACT_CLEAN["Is_Signed_Status"]
        .eq(False).fillna(False)
        & sign_dt.notna()
    )
)

# Điều kiện sử dụng thời điểm ký cho Conversion 30 ngày.
CONTRACT_CLEAN["Usable_Sign_Event"] = (
    CONTRACT_CLEAN["Is_Signed_Status"].fillna(False)
    & sign_dt.notna()
    & ~CONTRACT_CLEAN["Funnel_Order_Violation"]
).astype(bool)

# ============================================================
# STEP 3 — SANITY CHECK
# ============================================================
assert len(CONTRACT_CLEAN) == len(df_contract)
assert not (
    CONTRACT_CLEAN["Usable_Sign_Event"]
    & CONTRACT_CLEAN["Funnel_Order_Violation"]
).any()

step3 = pd.DataFrame({
    "Funnel Validation Metric": [
        "Apply > Approve",
        "Approve > Sign",
        "Apply > Sign",
        "Contracts with Any Order Violation",
        "Metadata-Defined Signed Status",
        "Unknown Credit Status",
        "Signed Status / Sign Time Conflict",
        "Usable Signed Contract Events",
    ],
    "Count": [
        CONTRACT_CLEAN["Apply_Approve_Violation"].sum(),
        CONTRACT_CLEAN["Approve_Sign_Violation"].sum(),
        CONTRACT_CLEAN["Apply_Sign_Violation"].sum(),
        CONTRACT_CLEAN["Funnel_Order_Violation"].sum(),
        CONTRACT_CLEAN["Is_Signed_Status"].eq(True).sum(),
        CONTRACT_CLEAN["Is_Signed_Status"].isna().sum(),
        CONTRACT_CLEAN["Signed_Time_Conflict_Flag"].sum(),
        CONTRACT_CLEAN["Usable_Sign_Event"].sum(),
    ],
})

bi_table(
    step3,
    "STEP 3 | CREDIT FUNNEL & SIGNED STATUS VALIDATION",
    note=(
        "Expected chronology: APPLY <= APPROVE <= SIGN; "
        "APPLY <= SIGN is checked independently.\n"
        "Violation = observed later-stage timestamp precedes an "
        "earlier-stage timestamp. Missing timestamps are not violations.\n"
        "Signed statuses: Active, Finished, Paid off, Signed, Written off. "
        "Others = not signed; unknown values remain unclassified.\n"
        "Usable Signed Event = signed status + valid sign timestamp "
        "+ no observed funnel-order violation. No contract is deleted."
    ),
    formats={"Count": "{:,.0f}"},
)
```

```python

# ============================================================
# STEP 4 — 30-DAY CONVERSION
# ============================================================
WINDOW_DAYS = 30
window = pd.Timedelta(days=WINDOW_DAYS)

# Mốc kết thúc quan sát chung cho hai nguồn.
lead_end = LEADS_CLEAN["DTIME_CREATED_VALID"].max()
contract_end = max(
    CONTRACT_CLEAN[f"{col}_VALID"].max()
    for col in contract_times
)

if pd.isna(lead_end) or pd.isna(contract_end):
    raise ValueError("Insufficient timestamps for a 30-day cohort.")

observation_end = min(lead_end, contract_end)

# Mỗi khách hàng chỉ có một lead đại diện:
# lead đầu tiên với creation timestamp hợp lệ.
lead_cohort = (
    LEADS_CLEAN.loc[
        LEADS_CLEAN["SKP_CLIENT"].notna()
        & LEADS_CLEAN["DTIME_CREATED_VALID"].notna(),
        ["SKP_CLIENT", "DTIME_CREATED_VALID", "Acceptance_Group"],
    ]
    .sort_values("DTIME_CREATED_VALID", kind="stable")
    .drop_duplicates("SKP_CLIENT", keep="first")
)

# Right-censoring: chỉ giữ lead đã đủ 30 ngày theo dõi.
eligible_cohort = (
    lead_cohort.loc[
        lead_cohort["DTIME_CREATED_VALID"].le(
            observation_end - window
        )
    ]
    .sort_values("DTIME_CREATED_VALID")
)

# Chỉ lấy các sự kiện ký có bằng chứng thời gian hợp lệ.
signed_events = (
    CONTRACT_CLEAN.loc[
        CONTRACT_CLEAN["Usable_Sign_Event"]
        & CONTRACT_CLEAN["SKP_CLIENT"].notna(),
        ["SKP_CLIENT", "SIGN_CONTRACT_TIME_VALID"],
    ]
    .sort_values("SIGN_CONTRACT_TIME_VALID")
)

# Forward as-of join: hợp đồng ký sớm nhất sau khi tạo lead,
# cùng khách hàng và không quá 30 ngày.
conversion_matched = pd.merge_asof(
    eligible_cohort,
    signed_events,
    left_on="DTIME_CREATED_VALID",
    right_on="SIGN_CONTRACT_TIME_VALID",
    by="SKP_CLIENT",
    direction="forward",
    tolerance=window,
)

conversion_matched["Converted_30D"] = (
    conversion_matched["SIGN_CONTRACT_TIME_VALID"].notna()
)

groups = ["Sentinel", "Real_Date", "Missing_Invalid"]

# Thống kê cohort trước và sau giới hạn thời gian theo dõi.
cohort_counts = (
    lead_cohort.groupby("Acceptance_Group")
    .size()
    .reindex(groups, fill_value=0)
)

conversion_counts = (
    conversion_matched.groupby("Acceptance_Group")
    .agg(
        Eligible_Customers=("SKP_CLIENT", "size"),
        Converted_Customers=("Converted_30D", "sum"),
    )
    .reindex(groups, fill_value=0)
)

conversion_summary = conversion_counts.assign(
    Cohort_Customers=cohort_counts,
    Conversion_Rate_30D_Pct=lambda d: (
        100 * d["Converted_Customers"]
        / d["Eligible_Customers"].replace(0, np.nan)
    ),
).reset_index()

conversion_summary = conversion_summary[
    [
        "Acceptance_Group",
        "Cohort_Customers",
        "Eligible_Customers",
        "Converted_Customers",
        "Conversion_Rate_30D_Pct",
    ]
]

# ============================================================
# STEP 4 — SANITY CHECK
# ============================================================
assert lead_cohort["SKP_CLIENT"].is_unique
assert conversion_matched["SKP_CLIENT"].is_unique
assert len(conversion_matched) == len(eligible_cohort)

converted = conversion_matched["Converted_30D"]

observed_gap = (
    conversion_matched.loc[converted, "SIGN_CONTRACT_TIME_VALID"]
    - conversion_matched.loc[converted, "DTIME_CREATED_VALID"]
)

assert observed_gap.between(pd.Timedelta(0), window).all()

assert (
    conversion_summary["Eligible_Customers"].sum()
    == len(eligible_cohort)
)

bi_table(
    conversion_summary,
    "STEP 4 | 30-DAY SIGNED CONTRACT CONVERSION",
    note=(
        "Formula: Conversion Rate (%) = Converted Customers "
        "/ Eligible Customers × 100.\n"
        "Anchor: earliest valid lead creation per SKP_CLIENT. "
        "Conversion: at least one usable signed contract within "
        "0–30 days after that anchor.\n"
        f"Common observation cutoff (UTC): {observation_end}. "
        "Eligible = lead created at least 30 days before cutoff.\n"
        "One customer contributes at most once. "
        "No lead-to-contract attribution is established; "
        "results are temporal associations, not causal effects.\n"
        "Cohort Customers exclude clients without a valid creation time. "
        "Undefined conversion rates have zero eligible customers."
    ),
    formats={
        "Cohort_Customers": "{:,.0f}",
        "Eligible_Customers": "{:,.0f}",
        "Converted_Customers": "{:,.0f}",
        "Conversion_Rate_30D_Pct": "{:.2f}%",
    },
)
```

```python

# ============================================================
# STEP 5 — FINAL VALIDATION
# ============================================================
raw_leads = len(df_lead)
raw_contracts = len(df_contract)

assert len(LEADS_CLEAN) == raw_leads
assert len(CONTRACT_CLEAN) == raw_contracts

assert set(df_lead.columns).issubset(LEADS_CLEAN.columns)
assert set(df_contract.columns).issubset(CONTRACT_CLEAN.columns)

# Bảo toàn khách hàng có định danh.
assert (
    df_lead["SKP_CLIENT"].nunique()
    == LEADS_CLEAN["SKP_CLIENT"].nunique()
)

assert (
    df_contract["SKP_CLIENT"].nunique()
    == CONTRACT_CLEAN["SKP_CLIENT"].nunique()
)

# Ba nhóm acceptance phải bao phủ toàn bộ lead.
assert LEADS_CLEAN["Acceptance_Group"].value_counts().sum() == raw_leads

# Bảo toàn cohort qua bước as-of join.
assert conversion_matched["SKP_CLIENT"].is_unique
assert len(conversion_matched) == len(eligible_cohort)

# Không phát sinh conversion ngoài 30 ngày theo dõi.
assert conversion_summary["Converted_Customers"].sum() <= len(
    eligible_cohort
)

# ============================================================
# REQUIRED OUTPUT — VALIDATION_SUMMARY
# ============================================================
VALIDATION_SUMMARY = pd.DataFrame({
    "Validation Metric": [
        "Raw Leads",
        "LEADS_CLEAN Rows",
        "Raw Contracts",
        "CONTRACT_CLEAN Rows",
        "Sentinel Leads",
        "Missing / Invalid Acceptance Dates",
        "Missing Actual Lead SLA",
        "Negative Lead SLA",
        "Contract Order Violations",
        "Signed Status / Time Conflicts",
        "Lead Cohort Customers",
        "Eligible 30-Day Customers",
        "Converted 30-Day Customers",
        "Lead Row Conservation",
        "Contract Row Conservation",
        "Customer Conservation",
        "Join Fan-out Check",
    ],
    "Value": [
        f"{raw_leads:,}",
        f"{len(LEADS_CLEAN):,}",
        f"{raw_contracts:,}",
        f"{len(CONTRACT_CLEAN):,}",
        f"{LEADS_CLEAN['Is_Sentinel_3000_Flag'].sum():,}",
        f"{LEADS_CLEAN['Acceptance_Group'].eq('Missing_Invalid').sum():,}",
        f"{LEADS_CLEAN['Actual_Lead_SLA'].isna().sum():,}",
        f"{LEADS_CLEAN['SLA_Negative_Flag'].sum():,}",
        f"{CONTRACT_CLEAN['Funnel_Order_Violation'].sum():,}",
        f"{CONTRACT_CLEAN['Signed_Time_Conflict_Flag'].sum():,}",
        f"{len(lead_cohort):,}",
        f"{len(eligible_cohort):,}",
        f"{conversion_matched['Converted_30D'].sum():,}",
        "PASS",
        "PASS",
        "PASS",
        "PASS",
    ],
})

bi_table(
    VALIDATION_SUMMARY,
    "FINAL VALIDATION | SENTINEL & CREDIT FUNNEL",
    note=(
        "Source rows and source column names are preserved in "
        "LEADS_CLEAN and CONTRACT_CLEAN.\n"
        "Sentinel values are retained as flags but excluded from "
        "accepted-date arithmetic. Funnel violations are flagged, "
        "not deleted.\n"
        "Conversion denominator: unique customers with one valid "
        "anchor lead and complete 30-day follow-up.\n"
        f"Observation cutoff (UTC): {observation_end}. "
        "The cutoff is inferred from observed source coverage; "
        "complete operational coverage must still be confirmed.\n"
        "PASS refers to programmatic integrity checks, not "
        "independent confirmation of all business assumptions."
    ),
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from html import escape
from textwrap import fill
from matplotlib import font_manager
from IPython.display import HTML, display

# ============================================================
# THEME & SLIDE-READY OUTPUT
# ============================================================
NAVY = "#17365D"
BLUE = "#85C1E9"
NEUTRAL = "#B4C7E7"
GRID = "#D9E2F3"

plt.rcParams.update({
    "font.family": (
        "Arial" if any(
            f.name == "Arial" for f in font_manager.fontManager.ttflist
        ) else "DejaVu Sans"
    ),
    "text.color": NAVY,
    "axes.labelcolor": NAVY,
    "xtick.color": NAVY,
    "ytick.color": NAVY,
})

def bi_table_21(df, title, question, note):
    styled = (
        df.style
        .hide(axis="index")
        .set_table_styles([
            {"selector": "th", "props": [
                ("background-color", NAVY), ("color", "white"),
                ("padding", "10px 14px"), ("text-align", "left"),
            ]},
            {"selector": "td", "props": [
                ("color", NAVY), ("padding", "9px 14px"),
                ("border-bottom", f"1px solid {GRID}"),
            ]},
        ])
    )
    display(HTML(
        f'<div style="font-family:Arial,sans-serif;color:{NAVY}">'
        f'<h3>{escape(title)}</h3>'
        f'<p><b>Question:</b> {escape(question)}</p>'
        f'{styled.to_html()}'
        f'<p style="font-size:12px;line-height:1.6">'
        f'{escape(note)}</p></div>'
    ))

def bi_bar_21(values, title, question, note, unit):
    values = values.dropna()
    if values.empty:
        bi_table_21(
            pd.DataFrame({"Status": ["No computable values"]}),
            title, question, note,
        )
        return

    fig, ax = plt.subplots(figsize=(12, 5))
    bars = ax.barh(
        values.index,
        values.to_numpy(),
        color=[NAVY] + [NEUTRAL] * (len(values) - 1),
        height=0.55,
    )
    ax.bar_label(
        bars,
        labels=[f"{v:,.2f}" if unit == "Days"
                else f"{v:,.0f}" for v in values],
        padding=5, fontsize=11, fontweight="bold", color=NAVY,
    )
    ax.invert_yaxis()
    ax.set_xlim(0, max(values.max(), 1) * 1.32)
    ax.set_xlabel(unit, fontweight="bold")
    ax.grid(axis="x", color=GRID, alpha=0.3)
    ax.set_axisbelow(True)
    ax.spines[["top", "right"]].set_visible(False)
    ax.set_title(
        title, loc="left", fontweight="bold",
        fontsize=15, color=NAVY, pad=18,
    )
    fig.text(
        0.04, 0.16, f"QUESTION: {question}",
        color=NAVY, fontsize=10, fontweight="bold",
    )
    fig.text(
        0.04, 0.045, fill(note, 115),
        color=NAVY, fontsize=9.5, va="bottom",
    )
    fig.subplots_adjust(
        left=0.32, right=0.93, top=0.83, bottom=0.29
    )
    plt.show()


# ============================================================
# STEP 1 — SCHEMA & SNAPSHOT
# ============================================================
required = {
    "df_customer": (df_customer, {"SKP_CLIENT"}),
    "df_events": (
        df_events,
        {"SKP_CLIENT", "SESSION_ID", "DTIME_EVENT_PARSED"},
    ),
    "LEADS_CLEAN": (
        LEADS_CLEAN, {"SKP_CLIENT", "DTIME_CREATED_VALID"},
    ),
    "df_call": (
        df_call,
        {
            "SKP_CLIENT", "TIME_OF_CALL_START",
            "TIME_OF_CALL_END", "CALL_RESULT_SYSTEM",
            "OFF_BY", "DURATIONS",
        },
    ),
    "df_valid_calls": (
        df_valid_calls,
        {
            "IS_EFFECTIVE_CALL", "CALL_RESULT_SYSTEM",
            "OFF_BY", "DURATIONS",
        },
    ),
}

for name, (frame, cols) in required.items():
    missing = cols - set(frame.columns)
    if missing:
        raise KeyError(f"{name}: missing {sorted(missing)}")

if df_customer["SKP_CLIENT"].isna().any():
    raise ValueError("CUSTOMER has missing client IDs.")

if not df_customer["SKP_CLIENT"].is_unique:
    raise ValueError("CUSTOMER must be unique by SKP_CLIENT.")

customer_base = df_customer[["SKP_CLIENT"]].copy()
customer_ids = pd.Index(customer_base["SKP_CLIENT"])

# Dùng timestamp đã parse từ các bước trước.
pb_time = pd.to_datetime(
    df_events["DTIME_EVENT_PARSED"], errors="coerce", utc=True
)
lead_time = pd.to_datetime(
    LEADS_CLEAN["DTIME_CREATED_VALID"], errors="coerce", utc=True
)

call_start = pd.to_datetime(
    df_call["TIME_OF_CALL_START"],
    errors="coerce", utc=True, format="mixed",
)
call_end = pd.to_datetime(
    df_call["TIME_OF_CALL_END"],
    errors="coerce", utc=True, format="mixed",
)

sources = {
    "PAST_BEHAVIOR": pb_time,
    "LEADS": lead_time,
    "CALL": call_start,
}

if any(s.notna().sum() == 0 for s in sources.values()):
    raise ValueError("A source has no valid event timestamp.")

# Chọn mốc chung thận trọng, ở độ chính xác timestamp.
# Không tự làm tròn lên cuối ngày để tránh lấy thêm tương lai.
T0 = min(s.max() for s in sources.values())
WINDOW_DAYS = 90
WINDOW_START = T0 - pd.Timedelta(days=WINDOW_DAYS)

coverage = pd.DataFrame({
    "Source": list(sources),
    "First Observed (UTC)": [
        str(s.min()) for s in sources.values()
    ],
    "Last Observed (UTC)": [
        str(s.max()) for s in sources.values()
    ],
    "Events After T0": [
        f"{s.gt(T0).sum():,}" for s in sources.values()
    ],
})

assert all(s.notna().any() for s in sources.values())
assert WINDOW_START < T0

bi_table_21(
    coverage,
    "STEP 1 | COMMON SNAPSHOT & DATA COVERAGE",
    "What historical cutoff avoids using later events?",
    (
        f"T0 = {T0}; Window = ({WINDOW_START}, {T0}]. "
        "T0 is the minimum of the latest observed timestamps "
        "across three sources. Counts after T0 refer to valid "
        "event timestamps. Source coverage continuity and "
        "source timezone require business verification."
    ),
)
```

```python

# ============================================================
# STEP 2A — VERIFIED EFFECTIVE CALL MAPPING (V2)
# ============================================================

# 1. Kiểm tra toàn bộ cột gốc có mặt trong dữ liệu đã xử lý.
source_cols = df_call.columns.tolist()

if not df_call.columns.is_unique:
    raise ValueError("Duplicate column names in df_call.")

missing = set(source_cols) - set(df_valid_calls.columns)
if missing:
    raise KeyError(f"Missing source columns: {sorted(missing)}")

# 2. Sử dụng trực tiếp IS_EFFECTIVE_CALL đã có từ V2.
reference_flag = pd.to_numeric(
    df_valid_calls["IS_EFFECTIVE_CALL"],
    errors="coerce",
)

if not reference_flag.dropna().isin([0, 1]).all():
    raise ValueError("IS_EFFECTIVE_CALL must contain 0/1/NA.")

# 3. Fingerprint theo tất cả cột nguồn, không chỉ CALL_ID.
ref_hash = pd.util.hash_pandas_object(
    df_valid_calls[source_cols],
    index=False,
)

raw_hash = pd.util.hash_pandas_object(
    df_call[source_cols],
    index=False,
)

reference_map = pd.DataFrame({
    "ROW_HASH": ref_hash.to_numpy(),
    "EFFECTIVE_FLAG": reference_flag.to_numpy(),
})

# Một bản ghi nguồn không được ánh xạ tới các nhãn mâu thuẫn.
if (
    reference_map.groupby("ROW_HASH")["EFFECTIVE_FLAG"]
    .nunique(dropna=False)
    .gt(1)
    .any()
):
    raise ValueError("Conflicting labels for identical row hashes.")

# Chỉ khử trùng khóa tra cứu, KHÔNG khử trùng call attempts.
label_lookup = (
    reference_map
    .drop_duplicates("ROW_HASH")
    .set_index("ROW_HASH")["EFFECTIVE_FLAG"]
)

raw_flag = raw_hash.map(label_lookup)

# Bảo đảm mọi reference đã gắn nhãn đều tìm thấy trong raw.
if not ref_hash.isin(raw_hash).all():
    raise ValueError(
        "Some V2 reference rows cannot be matched to raw CALL. "
        "Check source-column types and transformations."
    )

# 4. Outcome hợp lệ nếu có nhãn và cuộc gọi kết thúc trước T0.
outcome_known = (
    raw_flag.notna()
    & call_start.notna()
    & call_end.notna()
    & call_end.ge(call_start)
    & call_end.le(T0)
)

# Cuộc gọi hiệu quả: dùng nhãn V2, không tái suy diễn.
raw_effective = (
    raw_flag.eq(1).fillna(False) & outcome_known
)

# ============================================================
# SANITY CHECK
# ============================================================
assert len(raw_flag) == len(df_call)
assert not (raw_effective & ~outcome_known).any()
assert raw_effective.sum() <= outcome_known.sum()

matched = int(raw_flag.notna().sum())
unknown = len(df_call) - matched

mapping_summary = pd.DataFrame({
    "Validation Metric": [
        "Raw CALL Records",
        "V2 Reference Records",
        "Matched Raw Records",
        "Unmatched / Unknown Raw Records",
        "Mapping Coverage (%)",
        "CALL_ID Deduplication",
        "Label Rule Reconstruction",
    ],
    "Value": [
        f"{len(df_call):,}",
        f"{len(df_valid_calls):,}",
        f"{matched:,}",
        f"{unknown:,}",
        f"{matched / max(len(df_call), 1):.2%}",
        "NOT APPLIED",
        "NOT APPLIED",
    ],
})

bi_table_21(
    mapping_summary,
    "STEP 2A | EFFECTIVE CALL LABEL VALIDATION",
    "How much of the raw CALL population can be matched "
    "to existing V2 effective-call labels?",
    (
        "Matching key: fingerprint of all original CALL columns. "
        "Mapping Coverage = Matched Raw Records / Raw CALL Records. "
        "Unmatched records retain unknown effective-call status. "
        "No original CALL record is removed. "
        "Hash matching has a negligible but nonzero theoretical "
        "collision risk; upstream label definitions still require "
        "independent business validation."
    ),
)

# ============================================================
# STEP 2B — FILTER HISTORICAL EVENTS
# ============================================================
pb_mask = (
    pb_time.notna()
    & pb_time.le(T0)
    & df_events["SKP_CLIENT"].isin(customer_ids)
)

# Dựng lại thời điểm đầu/cuối của session từ log <= T0.
# Không dùng Session_End toàn kỳ nếu session đi qua T0.
pb_sessions_21 = (
    df_events.loc[
        pb_mask, ["SKP_CLIENT", "SESSION_ID"]
    ]
    .assign(Event_Time=pb_time.loc[pb_mask])
    .groupby("SESSION_ID", sort=False)
    .agg(
        SKP_CLIENT=("SKP_CLIENT", "first"),
        Session_Start=("Event_Time", "min"),
        Session_Last=("Event_Time", "max"),
    )
)

lead_mask = (
    lead_time.notna()
    & lead_time.le(T0)
    & LEADS_CLEAN["SKP_CLIENT"].isin(customer_ids)
)

lead_asof = (
    LEADS_CLEAN.loc[lead_mask, ["SKP_CLIENT"]]
    .assign(Lead_Time=lead_time.loc[lead_mask])
)

call_mask = (
    call_start.notna()
    & call_start.le(T0)
    & df_call["SKP_CLIENT"].isin(customer_ids)
)

call_asof = (
    df_call.loc[call_mask, ["SKP_CLIENT"]]
    .assign(
        Call_Start=call_start.loc[call_mask],
        Call_End=call_end.loc[call_mask],
        Outcome_Known=outcome_known.loc[call_mask],
        Verified_Effective=(
            raw_effective & outcome_known
        ).loc[call_mask],
    )
)

# ============================================================
# STEP 2C — CHANNEL RECENCY
# ============================================================
pb_last = (
    pb_sessions_21.groupby("SKP_CLIENT")["Session_Last"].max()
)
lead_last = lead_asof.groupby("SKP_CLIENT")["Lead_Time"].max()
call_last = (
    call_asof.loc[call_asof["Verified_Effective"]]
    .groupby("SKP_CLIENT")["Call_End"].max()
)

def recency_days(last_time):
    return (T0 - last_time).dt.total_seconds().div(86400)

pb_recency = recency_days(pb_last).rename(
    "PB_Session_Recency_Days"
)
lead_recency = recency_days(lead_last).rename(
    "Lead_Recency_Days"
)
call_recency = recency_days(call_last).rename(
    "Effective_Call_Recency_Days"
)

# Sanity check
assert all(
    s.dropna().ge(0).all()
    for s in [pb_recency, lead_recency, call_recency]
)
assert pb_sessions_21["Session_Last"].le(T0).all()
assert lead_asof["Lead_Time"].le(T0).all()
assert call_asof["Call_Start"].le(T0).all()

recency_medians = pd.Series({
    "PAST_BEHAVIOR Sessions": pb_recency.median(),
    "Lead Creation": lead_recency.median(),
    "Verified Effective Calls": call_recency.median(),
})

bi_bar_21(
    recency_medians,
    "STEP 2 | MULTI-CHANNEL RECENCY",
    "Which channel records the most recent customer activity?",
    (
        "Recency (days) = (T0 - Last Valid Interaction) / 86,400s. "
        "Lower = more recent. Medians exclude customers with no "
        "observed activity in the channel. "
        f"Eligible customers: PB={len(pb_recency):,}, "
        f"Leads={len(lead_recency):,}, Calls={len(call_recency):,}."
    ),
    "Days",
)
```

```python

# ============================================================
# STEP 3 — 90-DAY FREQUENCY
# ============================================================

# 1. PAST_BEHAVIOR: số phiên bắt đầu trong 90 ngày.
pb_90 = pb_sessions_21.loc[
    pb_sessions_21["Session_Start"].gt(WINDOW_START)
]

pb_frequency = (
    pb_90.groupby("SKP_CLIENT").size()
    .rename("PB_Session_Frequency_90D")
)

# Median khoảng cách giữa các session liên tiếp của cùng khách.
pb_sorted = pb_90.sort_values(
    ["SKP_CLIENT", "Session_Start"], kind="stable"
)

session_gaps = (
    pb_sorted.groupby("SKP_CLIENT")["Session_Start"]
    .diff()
    .dt.total_seconds()
    .div(86400)
)

pb_gap = (
    session_gaps.groupby(pb_sorted["SKP_CLIENT"])
    .median()
    .rename("Median_Inter_Session_Gap_Days_90D")
)

# 2. LEADS: đếm số bản ghi Lead được tạo trong 90 ngày.
lead_90 = lead_asof.loc[
    lead_asof["Lead_Time"].gt(WINDOW_START)
]

lead_frequency = (
    lead_90.groupby("SKP_CLIENT").size()
    .rename("Lead_Frequency_90D")
)

# 3. CALL: đếm từng attempt, không deduplicate CALL_ID.
call_90 = call_asof.loc[
    call_asof["Call_Start"].gt(WINDOW_START)
]

call_frequency = (
    call_90.groupby("SKP_CLIENT").agg(
        Total_Call_Attempts_90D=("Call_Start", "size"),
        Known_Call_Outcomes_90D=("Outcome_Known", "sum"),
        Effective_Connected_Calls_90D=(
            "Verified_Effective", "sum"
        ),
    )
)

# ============================================================
# SANITY CHECK
# ============================================================
assert pb_frequency.sum() == len(pb_90)
assert lead_frequency.sum() == len(lead_90)
assert (
    call_frequency["Total_Call_Attempts_90D"].sum()
    == len(call_90)
)

assert (
    call_frequency["Effective_Connected_Calls_90D"]
    <= call_frequency["Known_Call_Outcomes_90D"]
).all()

assert pb_gap.dropna().ge(0).all()

# ============================================================
# SLIDE-READY OUTPUT
# ============================================================
volume_90d = pd.Series({
    "Call Attempts": len(call_90),
    "Verified Effective Calls": int(
        call_90["Verified_Effective"].sum()
    ),
    "PAST_BEHAVIOR Sessions": len(pb_90),
    "Leads Created": len(lead_90),
})

bi_bar_21(
    volume_90d,
    "STEP 3 | 90-DAY MULTI-CHANNEL INTERACTION VOLUME",
    "Does observed contact effort translate into effective interaction?",
    (
        "Frequency = number of observed records or sessions "
        "within the same 90-day window. "
        "PB = sessions started; Leads = creation records; "
        "Call Attempts = raw call rows started. "
        "Effective Calls are a subset of Attempts. "
        "The four bars are not mutually exclusive."
    ),
    "Events / Sessions / Calls",
)
```

```python

# ============================================================
# STEP 4 — AGGREGATE EACH SOURCE
# ============================================================
pb_features = pd.concat(
    [pb_recency, pb_frequency, pb_gap], axis=1
)

lead_features = pd.concat(
    [lead_recency, lead_frequency], axis=1
)

call_features = pd.concat(
    [call_recency, call_frequency], axis=1
)

# Mỗi nguồn phải có một dòng trên SKP_CLIENT.
assert all(
    frame.index.is_unique
    for frame in [pb_features, lead_features, call_features]
)

# ============================================================
# CUSTOMER LEFT JOIN — NO FAN-OUT
# ============================================================
CUSTOMER_MULTICHANNEL_FEATURES = (
    customer_base
    .merge(
        pb_features,
        left_on="SKP_CLIENT",
        right_index=True,
        how="left",
        validate="one_to_one",
    )
    .merge(
        lead_features,
        left_on="SKP_CLIENT",
        right_index=True,
        how="left",
        validate="one_to_one",
    )
    .merge(
        call_features,
        left_on="SKP_CLIENT",
        right_index=True,
        how="left",
        validate="one_to_one",
    )
)

features = CUSTOMER_MULTICHANNEL_FEATURES

frequency_cols = [
    "PB_Session_Frequency_90D",
    "Lead_Frequency_90D",
    "Total_Call_Attempts_90D",
    "Known_Call_Outcomes_90D",
    "Effective_Connected_Calls_90D",
]

# Frequency = 0 nếu không ghi nhận hoạt động trong 90D.
# Không thay NaN của Recency hoặc Median Gap.
features[frequency_cols] = (
    features[frequency_cols].fillna(0).astype("int32")
)

# Unknown outcomes không được coi là cuộc gọi thất bại.
features["Unknown_Call_Outcomes_90D"] = (
    features["Total_Call_Attempts_90D"]
    - features["Known_Call_Outcomes_90D"]
)

# Effective / Attempts.
# Nếu còn unknown outcomes hoặc không có attempts -> NaN.
features["Effective_Connection_Rate_90D"] = (
    features["Effective_Connected_Calls_90D"]
    .div(
        features["Total_Call_Attempts_90D"]
        .replace(0, np.nan)
    )
    .where(features["Unknown_Call_Outcomes_90D"].eq(0))
)

# ============================================================
# SANITY CHECK
# ============================================================
n_customers = len(customer_base)

assert len(features) == n_customers
assert features["SKP_CLIENT"].is_unique

assert features["Unknown_Call_Outcomes_90D"].ge(0).all()
assert (
    features["Effective_Connected_Calls_90D"]
    <= features["Known_Call_Outcomes_90D"]
).all()

assert (
    features["Effective_Connection_Rate_90D"]
    .dropna().between(0, 1).all()
)

assert features["PB_Session_Frequency_90D"].sum() == len(pb_90)
assert features["Lead_Frequency_90D"].sum() == len(lead_90)
assert features["Total_Call_Attempts_90D"].sum() == len(call_90)

# ============================================================
# SLIDE-READY VALIDATION
# ============================================================
join_summary = pd.DataFrame({
    "Integrity Metric": [
        "CUSTOMER Input Rows",
        "Final Customer Rows",
        "JOIN Fan-out",
        "Call Attempts in 90D",
        "Known Call Outcomes in 90D",
        "Unknown Call Outcomes in 90D",
        "Effective Connections in 90D",
    ],
    "Count": [
        n_customers,
        len(features),
        len(features) - n_customers,
        len(call_90),
        int(call_90["Outcome_Known"].sum()),
        int((~call_90["Outcome_Known"]).sum()),
        int(call_90["Verified_Effective"].sum()),
    ],
})

join_summary["Count"] = join_summary["Count"].map(
    lambda x: f"{x:,}"
)

bi_table_21(
    join_summary,
    "STEP 4 | MULTI-SOURCE CUSTOMER INTEGRITY",
    "Are contact pressure and verified engagement measured consistently?",
    (
        "Grain: one row per SKP_CLIENT. "
        "Each source is aggregated independently before LEFT JOIN. "
        "Effective Connection Rate = Effective Calls / Total Attempts. "
        "A rate is undefined when no attempts exist or any "
        "attempt has an unverified outcome. "
        "No full-year 2025 app activity is used."
    ),
)
```

```python

# ============================================================
# STEP 5 — FINAL VALIDATION
# ============================================================

recency_cols = [
    "PB_Session_Recency_Days",
    "Lead_Recency_Days",
    "Effective_Call_Recency_Days",
    "Median_Inter_Session_Gap_Days_90D",
]

assert features[frequency_cols].ge(0).all().all()
assert features[recency_cols].stack().ge(0).all()

assert features["SKP_CLIENT"].nunique() == n_customers

assert (
    features["Effective_Connected_Calls_90D"].sum()
    == int(call_90["Verified_Effective"].sum())
)

assert (
    features["Unknown_Call_Outcomes_90D"].sum()
    == int((~call_90["Outcome_Known"]).sum())
)

assert call_asof.loc[
    call_asof["Outcome_Known"], "Call_End"
].le(T0).all()

# ============================================================
# VALIDATION SUMMARY
# ============================================================
VALIDATION_SUMMARY_21 = pd.DataFrame({
    "Validation Metric": [
        "Snapshot T0 (UTC)",
        "Frequency Window",
        "Raw CUSTOMER Rows",
        "Final Feature Rows",
        "JOIN Fan-out Rows",
        "PAST_BEHAVIOR Sessions (90D)",
        "Leads Created (90D)",
        "Total Call Attempts (90D)",
        "Verified Effective Calls (90D)",
        "Unknown Call Outcomes (90D)",
        "Customers with Computable Connection Rate",
        "Customers with Computable Median Session Gap",
        "Effective Rule Verification",
        "Customer Conservation",
        "Future Event Exclusion",
    ],
    "Result": [
        str(T0),
        f"{WINDOW_DAYS} days",
        f"{n_customers:,}",
        f"{len(features):,}",
        f"{len(features) - n_customers:,}",
        f"{len(pb_90):,}",
        f"{len(lead_90):,}",
        f"{len(call_90):,}",
        f"{call_90['Verified_Effective'].sum():,}",
        f"{(~call_90['Outcome_Known']).sum():,}",
        f"{features['Effective_Connection_Rate_90D'].notna().sum():,}",
        f"{features['Median_Inter_Session_Gap_Days_90D'].notna().sum():,}",
        "PASS",
        "PASS",
        "PASS",
    ],
})

bi_table_21(
    VALIDATION_SUMMARY_21,
    "FINAL VALIDATION | MULTI-CHANNEL ENGAGEMENT",
    "Are the engineered features consistent, complete and leakage-controlled?",
    (
        "Recency = (T0 - Last Valid Interaction) / 86,400 seconds. "
        "Frequency = observed events or sessions in the preceding 90 days. "
        "Median Gap = median interval between consecutive PB session starts. "
        "CALL_ID is not deduplicated. PASS confirms programmatic "
        "checks, not full business validation of source coverage."
    ),
)

# ============================================================
# MAXIMUM TWO BUSINESS INSIGHTS
# ============================================================

# Nhóm có đủ outcome để đánh giá áp lực liên hệ.
complete_calls = (
    features["Total_Call_Attempts_90D"].gt(0)
    & features["Unknown_Call_Outcomes_90D"].eq(0)
)

p75_pressure = features.loc[
    complete_calls, "Total_Call_Attempts_90D"
].quantile(0.75)

high_pressure_low_response = (
    complete_calls
    & features["Total_Call_Attempts_90D"].ge(p75_pressure)
    & features["Effective_Connected_Calls_90D"].eq(0)
)

high_pressure_n = int(high_pressure_low_response.sum())
eligible_calls_n = int(complete_calls.sum())

# Nhóm có hoạt động PB nhưng không ghi nhận attempt gọi.
pb_active = features["PB_Session_Frequency_90D"].gt(0)

digital_without_calls = (
    pb_active
    & features["Total_Call_Attempts_90D"].eq(0)
)

digital_n = int(digital_without_calls.sum())
pb_active_n = int(pb_active.sum())

insight_rows = []

if eligible_calls_n > 0:
    insight_rows.append({
        "Business Insight": "High Contact Pressure / Zero Effective Calls",
        "Statistical Evidence": (
            f"{high_pressure_n:,} / {eligible_calls_n:,} "
            f"customers ({high_pressure_n / eligible_calls_n:.2%}); "
            f"attempts >= P75 ({p75_pressure:.2f})"
        ),
        "Recommended Action": (
            "Pilot lower redial intensity and alternative "
            "contact timing; monitor verified connection lift."
        ),
    })

if pb_active_n > 0:
    insight_rows.append({
        "Business Insight": "Recorded PB Activity Without Call Attempts",
        "Statistical Evidence": (
            f"{digital_n:,} / {pb_active_n:,} "
            f"PB-active customers ({digital_n / pb_active_n:.2%})"
        ),
        "Recommended Action": (
            "Test digital-first follow-up and measure "
            "subsequent qualified engagement."
        ),
    })

INSIGHT_SUMMARY_21 = pd.DataFrame(insight_rows)

if not INSIGHT_SUMMARY_21.empty:
    bi_table_21(
        INSIGHT_SUMMARY_21,
        "BUSINESS INSIGHTS | ENGAGEMENT VS CONTACT PRESSURE",
        "Where should outreach effort be optimized?",
        (
            "Insight 1 denominator: customers with >= 1 call "
            "attempt and fully verified outcomes. P75 is "
            "calculated within this population. "
            "Insight 2 denominator: customers with >= 1 "
            "recorded PB session in 90 days. "
            "These observations suggest experiments, "
            "not causal conclusions."
        ),
    )
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from html import escape
from IPython.display import HTML, display

# ============================================================
# 1. BI THEME & HTML OUTPUT
# ============================================================
NAVY = "#17365D"
BLUE = "#85C1E9"
NEUTRAL = "#B4C7E7"
GRID = "#D9E2F3"

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "text.color": NAVY,
    "axes.labelcolor": NAVY,
    "xtick.color": NAVY,
    "ytick.color": NAVY,
})

def show_bi(data, title, question, note):
    style = (
        data.style
        .format(na_rep="—")
        .hide(axis="index")
        .set_table_styles([
            {"selector": "th", "props": [
                ("background-color", NAVY),
                ("color", "white"),
                ("padding", "10px 14px"),
                ("text-align", "left"),
            ]},
            {"selector": "td", "props": [
                ("color", NAVY),
                ("padding", "9px 14px"),
                ("border-bottom", f"1px solid {GRID}"),
            ]},
        ])
    )

    display(HTML(
        f'<div style="font-family:Arial,sans-serif;color:{NAVY}">'
        f'<h3 style="font-size:17px">{escape(title)}</h3>'
        f'<p><b>Question:</b> {escape(question)}</p>'
        f'{style.to_html()}'
        f'<p style="font-size:12px;line-height:1.6">'
        f'{escape(note)}</p></div>'
    ))

# ============================================================
# 2. VALIDATE REUSED FEATURES
# ============================================================
feature_cols = [
    "SKP_CLIENT",
    "PB_Session_Frequency_90D",
    "Total_Call_Attempts_90D",
    "Effective_Connected_Calls_90D",
    "Unknown_Call_Outcomes_90D",
    "Effective_Connection_Rate_90D",
]

missing_features = (
    set(feature_cols) - set(CUSTOMER_MULTICHANNEL_FEATURES.columns)
)
missing_customer = {
    "SKP_CLIENT", "AVG_SESSION_PER_WEEK_2025"
} - set(df_customer.columns)

if missing_features or missing_customer:
    raise KeyError(
        f"Missing features: {sorted(missing_features)}; "
        f"Missing CUSTOMER: {sorted(missing_customer)}"
    )

call_cols = {
    "SKP_CLIENT", "CALL_TYPE", "TIME_OF_CALL_START",
    "TIME_OF_CALL_END", "DURATIONS", "CALL_RESULT_SYSTEM",
}

if not call_cols.issubset(df_call.columns):
    raise KeyError(sorted(call_cols - set(df_call.columns)))

if not CUSTOMER_MULTICHANNEL_FEATURES["SKP_CLIENT"].is_unique:
    raise ValueError("Customer-level features must be unique.")

if not df_customer["SKP_CLIENT"].is_unique:
    raise ValueError("CUSTOMER must have unique SKP_CLIENT.")

if WINDOW_START != T0 - pd.Timedelta(days=90):
    raise ValueError("Inconsistent 90-day window from Action 2.1.")

# ============================================================
# 3. REUSE FEATURES + ANNUAL APP DESCRIPTIVE VARIABLE
# ============================================================
CUSTOMER_FREQUENCY_TABLE = (
    CUSTOMER_MULTICHANNEL_FEATURES[feature_cols]
    .merge(
        df_customer[
            ["SKP_CLIENT", "AVG_SESSION_PER_WEEK_2025"]
        ],
        on="SKP_CLIENT",
        how="left",
        validate="one_to_one",
    )
)

CUSTOMER_FREQUENCY_TABLE["AVG_SESSION_PER_WEEK_2025"] = (
    pd.to_numeric(
        CUSTOMER_FREQUENCY_TABLE["AVG_SESSION_PER_WEEK_2025"],
        errors="coerce",
    )
    .replace([np.inf, -np.inf], np.nan)
)

# ============================================================
# 4. PREPARE VALID OUTBOUND HISTORY
# ============================================================
# Cần thêm 7 ngày trước cửa sổ 90D để tính prior calls
# cho những cuộc gọi ở đầu cửa sổ.
LOOKBACK_START = WINDOW_START - pd.Timedelta(days=7)

start = pd.to_datetime(
    df_call["TIME_OF_CALL_START"],
    errors="coerce", utc=True, format="mixed",
)

end = pd.to_datetime(
    df_call["TIME_OF_CALL_END"],
    errors="coerce", utc=True, format="mixed",
)

duration = pd.to_numeric(
    df_call["DURATIONS"], errors="coerce"
).replace([np.inf, -np.inf], np.nan)

outbound_mask = (
    df_call["CALL_TYPE"]
    .astype("string").str.strip().str.casefold()
    .eq("outbound").fillna(False)
)

valid_mask = (
    outbound_mask
    & df_call["SKP_CLIENT"].notna()
    & start.between(LOOKBACK_START, T0)
    & end.ge(start)
    & duration.ge(0)
)

outbound_history = (
    df_call.loc[
        valid_mask,
        ["SKP_CLIENT", "CALL_RESULT_SYSTEM"],
    ]
    .assign(
        Call_Start=start.loc[valid_mask],
        Call_End=end.loc[valid_mask],
        One=np.int8(1),
    )
    .sort_values(
        ["SKP_CLIENT", "Call_Start"],
        kind="stable",
        ignore_index=True,
    )
)

# ============================================================
# 5. SANITY CHECK & OUTPUT
# ============================================================
assert len(CUSTOMER_FREQUENCY_TABLE) == len(df_customer)
assert CUSTOMER_FREQUENCY_TABLE["SKP_CLIENT"].is_unique
assert outbound_history["Call_Start"].le(T0).all()
assert outbound_history["Call_End"].ge(
    outbound_history["Call_Start"]
).all()

if outbound_history.empty:
    raise ValueError("No valid Outbound calls in the analysis period.")

step1_summary = pd.DataFrame({
    "Validation Metric": [
        "Customer Rows Preserved",
        "Snapshot T0 (UTC)",
        "Frequency Window",
        "Valid Outbound History Records",
        "Annual App Variable",
    ],
    "Result": [
        f"{len(CUSTOMER_FREQUENCY_TABLE):,}",
        str(T0),
        "90 days",
        f"{len(outbound_history):,}",
        "Descriptive only",
    ],
})

show_bi(
    step1_summary,
    "STEP 1 | HISTORICAL FREQUENCY & SOURCE VALIDATION",
    "Can frequency and outbound call pressure be analyzed "
    "without leaking future information?",
    (
        f"Snapshot: {T0}. Frequency window: ({WINDOW_START}, {T0}]. "
        "Call history additionally includes seven preceding days. "
        "AVG_SESSION_PER_WEEK_2025 summarizes the full calendar year "
        "and is excluded from historical segmentation and scoring. "
        "No CALL_ID deduplication is applied."
    ),
)
```

```python

# ============================================================
# STEP 2 — OUTBOUND FREQUENCY & CUSTOMER SEGMENTATION
# ============================================================

# Đếm Outbound Attempts trong đúng cửa sổ 90 ngày.
outbound_90_mask = (
    outbound_history["Call_Start"] > WINDOW_START
)

outbound_frequency = (
    outbound_history.loc[outbound_90_mask]
    .groupby("SKP_CLIENT", sort=False)
    .size()
    .rename("Outbound_Call_Attempts_90D")
)

CUSTOMER_FREQUENCY_TABLE = (
    CUSTOMER_FREQUENCY_TABLE.merge(
        outbound_frequency,
        left_on="SKP_CLIENT",
        right_index=True,
        how="left",
        validate="one_to_one",
    )
)

customers = CUSTOMER_FREQUENCY_TABLE

customers["Outbound_Call_Attempts_90D"] = (
    customers["Outbound_Call_Attempts_90D"]
    .fillna(0)
    .astype("int32")
)

# Digital Activity: chỉ dùng PB frequency đã nghiệm thu.
customers["Digital_Activity_90D"] = np.where(
    customers["PB_Session_Frequency_90D"].gt(0),
    "PB Activity Observed",
    "No PB Activity Observed",
)

# Responsiveness: giữ riêng nhóm outcome chưa đầy đủ.
attempts = customers["Total_Call_Attempts_90D"]
effective = customers["Effective_Connected_Calls_90D"]
unknown = customers["Unknown_Call_Outcomes_90D"]

customers["Call_Responsiveness_90D"] = np.select(
    [
        attempts.eq(0),
        unknown.gt(0),
        effective.gt(0),
    ],
    [
        "No Call Attempts",
        "Incomplete Outcomes",
        "Effective Connection",
    ],
    default="No Effective Connection",
)

# Contact Pressure chỉ dùng Outbound, không dùng Inbound.
outbound_count = customers["Outbound_Call_Attempts_90D"]
contacted = outbound_count.loc[outbound_count.gt(0)]

p75_outbound = contacted.quantile(0.75)
has_pressure_variation = contacted.nunique() > 1

customers["Call_Pressure_90D"] = np.select(
    [
        outbound_count.eq(0),
        outbound_count.ge(p75_outbound)
        & has_pressure_variation,
    ],
    ["No Outbound Contact", "High Relative Pressure"],
    default=(
        "Below P75 Pressure"
        if has_pressure_variation
        else "No Pressure Separation"
    ),
)

# ============================================================
# SANITY CHECK
# ============================================================
assert len(customers) == len(df_customer)
assert customers["SKP_CLIENT"].is_unique

assert (
    customers["Outbound_Call_Attempts_90D"].sum()
    == outbound_90_mask.sum()
)

assert effective.le(attempts).all()
assert unknown.le(attempts).all()

# Phân bố cả ba chiều; App 2025 chỉ thống kê mô tả.
segment_cols = [
    "Digital_Activity_90D",
    "Call_Responsiveness_90D",
    "Call_Pressure_90D",
]

segments = (
    customers[
        segment_cols + ["AVG_SESSION_PER_WEEK_2025"]
    ]
    .melt(
        id_vars="AVG_SESSION_PER_WEEK_2025",
        value_vars=segment_cols,
        var_name="Dimension",
        value_name="Segment",
    )
    .groupby(["Dimension", "Segment"], sort=False)
    .agg(
        Customers=("Segment", "size"),
        Median_App_Sessions_2025=(
            "AVG_SESSION_PER_WEEK_2025", "median"
        ),
    )
    .reset_index()
)

assert (
    segments.groupby("Dimension")["Customers"].sum()
    .eq(len(customers)).all()
)

segments["Median_App_Sessions_2025"] = (
    segments["Median_App_Sessions_2025"].round(2)
)

show_bi(
    segments,
    "STEP 2 | ENGAGEMENT VS CONTACT PRESSURE",
    "Which customers show genuine recorded activity, "
    "and which are primarily exposed to outbound contact?",
    (
        "Digital Activity = PB Session Frequency > 0 in 90D. "
        "Call Responsiveness uses verified features from Action 2.1. "
        f"High Relative Pressure = Outbound Calls >= P75 "
        f"({p75_outbound:.2f}, if variation exists). "
        "Annual App Sessions is an ex-post 2025 descriptive metric only. "
        "Each dimension independently sums to the customer population."
    ),
)
```

```python

# ============================================================
# STEP 3 — PRIOR OUTBOUND CALLS IN 7 DAYS
# ============================================================

# Rolling window = [Current Call Time - 7D, Current Call Time)
# Không dùng cuộc gọi hiện tại hoặc cuộc gọi tương lai.
prior_counts = (
    outbound_history
    .groupby("SKP_CLIENT", sort=False)
    .rolling(
        "7D",
        on="Call_Start",
        closed="left",
    )["One"]
    .sum()
)

outbound_history["Prior_Outbound_Calls_7D"] = (
    prior_counts.to_numpy()
)

outbound_history["Prior_Outbound_Calls_7D"] = (
    outbound_history["Prior_Outbound_Calls_7D"]
    .fillna(0)
    .astype("int32")
)

# Chỉ đánh giá Answer Rate trên calls bắt đầu trong 90D.
outbound_90 = outbound_history.loc[
    outbound_history["Call_Start"].gt(WINDOW_START)
].copy()

# Quan sát kết quả hệ thống, không diễn giải RESPONSE_RESULT_2.
result = (
    outbound_90["CALL_RESULT_SYSTEM"]
    .astype("string")
    .str.strip()
    .str.upper()
    .replace(["", "NAN", "NULL", "NONE", "N/A"], pd.NA)
)

# Chỉ dùng kết quả có thể quan sát tại hoặc trước T0.
outbound_90["Outcome_Observed"] = (
    result.notna()
    & outbound_90["Call_End"].le(T0)
)

outbound_90["Recorded_ANSWER"] = (
    outbound_90["Outcome_Observed"]
    & result.eq("ANSWER").fillna(False)
)

# Bins chỉ để trực quan hóa cường độ.
bin_labels = ["0", "1", "2–3", "4–7", "8–15", "16+"]

outbound_90["Prior_7D_Bin"] = pd.cut(
    outbound_90["Prior_Outbound_Calls_7D"],
    bins=[-1, 0, 1, 3, 7, 15, np.inf],
    labels=bin_labels,
)

CALL_FATIGUE_ANALYSIS = (
    outbound_90
    .groupby("Prior_7D_Bin", observed=False)
    .agg(
        Outbound_Attempts=("One", "sum"),
        Observed_Outcomes=("Outcome_Observed", "sum"),
        ANSWER_Count=("Recorded_ANSWER", "sum"),
    )
    .reset_index()
)

# ANSWER Rate theo từng mức cường độ gọi.
CALL_FATIGUE_ANALYSIS["Answer_Rate_Pct"] = (
    100
    * CALL_FATIGUE_ANALYSIS["ANSWER_Count"]
    / CALL_FATIGUE_ANALYSIS["Observed_Outcomes"]
    .replace(0, np.nan)
)

# ============================================================
# SANITY CHECK
# ============================================================
assert len(outbound_90) == outbound_90_mask.sum()
assert outbound_90["Prior_Outbound_Calls_7D"].ge(0).all()

assert (
    CALL_FATIGUE_ANALYSIS["Outbound_Attempts"].sum()
    == len(outbound_90)
)

assert (
    CALL_FATIGUE_ANALYSIS["ANSWER_Count"]
    <= CALL_FATIGUE_ANALYSIS["Observed_Outcomes"]
).all()

assert (
    CALL_FATIGUE_ANALYSIS["Observed_Outcomes"]
    <= CALL_FATIGUE_ANALYSIS["Outbound_Attempts"]
).all()

# ============================================================
# ANSWER RATE CURVE — SLIDE-READY
# ============================================================
curve = CALL_FATIGUE_ANALYSIS.loc[
    CALL_FATIGUE_ANALYSIS["Observed_Outcomes"].gt(0)
]

if curve.empty:
    raise ValueError("No observed outbound outcomes for Answer Rate.")

fig, ax = plt.subplots(figsize=(12, 5.5))

x = np.arange(len(curve))
y = curve["Answer_Rate_Pct"].to_numpy()

ax.plot(
    x, y,
    color=NAVY,
    linewidth=2.5,
    marker="o",
    markersize=8,
)

for xi, yi in zip(x, y):
    ax.annotate(
        f"{yi:.2f}%",
        (xi, yi),
        xytext=(0, 12),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold",
        color=NAVY,
    )

ax.set_xticks(x)
ax.set_xticklabels([
    f"{label}\n(n={n:,})"
    for label, n in zip(
        curve["Prior_7D_Bin"],
        curve["Observed_Outcomes"],
    )
])

ax.set_ylim(0, 110)
ax.set_ylabel("Recorded ANSWER Rate (%)", fontweight="bold")
ax.set_xlabel("Prior Outbound Calls in 7 Days", fontweight="bold")

ax.set_title(
    "CALL FATIGUE | 7-DAY OUTBOUND INTENSITY VS ANSWER RATE",
    loc="left", fontsize=15, fontweight="bold",
    color=NAVY, pad=20,
)

ax.grid(axis="y", color=GRID, alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)

fig.text(
    0.07, 0.16,
    "QUESTION: Does answer rate decrease as prior outbound "
    "call intensity increases?",
    color=NAVY, fontsize=10, fontweight="bold",
)

fig.text(
    0.07, 0.045,
    "Prior Calls = number of earlier outbound starts in [t-7D, t). "
    "ANSWER Rate = ANSWER-coded calls / calls with observed system "
    "outcomes × 100.\n"
    "n = observed outcomes per bin. "
    "Unknown outcomes are excluded from the rate. "
    "Observed association is not causation.",
    color=NAVY, fontsize=9,
    linespacing=1.5,
)

fig.subplots_adjust(
    left=0.09, right=0.96,
    top=0.83, bottom=0.32,
)

plt.show()
```

```python

# ============================================================
# STEP 4 — CALL FATIGUE COMPARISON
# ============================================================
observed = outbound_90.loc[
    outbound_90["Outcome_Observed"]
]

prior = observed["Prior_Outbound_Calls_7D"]
positive_prior = prior.loc[prior.gt(0)]

# P75 được xác định từ phân phối thực tế.
high_threshold = (
    int(np.ceil(positive_prior.quantile(0.75)))
    if not positive_prior.empty else None
)

baseline_mask = prior.eq(0)
high_mask = (
    prior.ge(high_threshold)
    if high_threshold is not None
    else pd.Series(False, index=observed.index)
)

baseline_n = int(baseline_mask.sum())
high_n = int(high_mask.sum())

baseline_answer = int(
    observed.loc[baseline_mask, "Recorded_ANSWER"].sum()
)
high_answer = int(
    observed.loc[high_mask, "Recorded_ANSWER"].sum()
)

baseline_rate = (
    100 * baseline_answer / baseline_n
    if baseline_n else np.nan
)
high_rate = (
    100 * high_answer / high_n
    if high_n else np.nan
)

delta_pp = high_rate - baseline_rate

# Ngưỡng tối thiểu chỉ để hạn chế diễn giải mẫu quá nhỏ,
# không phải kiểm định ý nghĩa thống kê.
MIN_OUTCOMES = 100

sufficient = (
    baseline_n >= MIN_OUTCOMES
    and high_n >= MIN_OUTCOMES
)

observed_decline = (
    sufficient
    and np.isfinite(delta_pp)
    and delta_pp < 0
)

fatigue_comparison = pd.DataFrame({
    "Pressure Group": [
        "No Prior Outbound Calls",
        "High Prior Outbound Pressure",
    ],
    "Observed Outcomes": [baseline_n, high_n],
    "ANSWER Count": [baseline_answer, high_answer],
    "Answer Rate (%)": [
        round(baseline_rate, 2),
        round(high_rate, 2),
    ],
})

show_bi(
    fatigue_comparison,
    "STEP 4 | HIGH VS ZERO PRIOR OUTBOUND PRESSURE",
    "Is the recorded ANSWER rate lower under higher "
    "recent outbound contact intensity?",
    (
        f"High Pressure = prior 7-day calls >= P75 "
        f"({high_threshold if high_threshold is not None else 'N/A'}). "
        f"Answer Rate Difference = {delta_pp:.2f} percentage points. "
        f"Observed Decline with adequate sample: "
        f"{'YES' if observed_decline else 'NOT ESTABLISHED'}. "
        f"Each group requires at least {MIN_OUTCOMES:,} "
        "observed outcomes for descriptive comparison. "
        "This is not a causal or cluster-adjusted statistical test."
    ),
)

# ============================================================
# CUSTOMER-LEVEL OUTBOUND RESPONSE
# ============================================================
outbound_response = (
    outbound_90.groupby("SKP_CLIENT", sort=False)
    .agg(
        Observed_Outbound_Outcomes_90D=(
            "Outcome_Observed", "sum"
        ),
        Recorded_ANSWER_Calls_90D=(
            "Recorded_ANSWER", "sum"
        ),
    )
)

CUSTOMER_FREQUENCY_TABLE = (
    customers.merge(
        outbound_response,
        left_on="SKP_CLIENT",
        right_index=True,
        how="left",
        validate="one_to_one",
    )
)

customers = CUSTOMER_FREQUENCY_TABLE

new_counts = [
    "Observed_Outbound_Outcomes_90D",
    "Recorded_ANSWER_Calls_90D",
]

customers[new_counts] = (
    customers[new_counts].fillna(0).astype("int32")
)

# Mục tiêu: khách chịu áp lực Outbound cao,
# có đủ kết quả quan sát và không ghi nhận ANSWER.
priority_mask = (
    customers["Call_Pressure_90D"]
    .eq("High Relative Pressure")
    & customers["Outbound_Call_Attempts_90D"].gt(0)
    & customers["Observed_Outbound_Outcomes_90D"].eq(
        customers["Outbound_Call_Attempts_90D"]
    )
    & customers["Recorded_ANSWER_Calls_90D"].eq(0)
)

customers["Next_Best_Action_22"] = pd.Series(
    pd.NA, index=customers.index, dtype="string"
)

customers.loc[
    priority_mask, "Next_Best_Action_22"
] = (
    "Pilot alternative contact timing and lower redial intensity"
    if observed_decline
    else "Review contact timing and test alternative outreach"
)

# ============================================================
# SANITY CHECK
# ============================================================
assert len(customers) == len(df_customer)
assert customers["SKP_CLIENT"].is_unique

assert (
    customers["Recorded_ANSWER_Calls_90D"]
    <= customers["Observed_Outbound_Outcomes_90D"]
).all()

assert (
    customers["Observed_Outbound_Outcomes_90D"]
    <= customers["Outbound_Call_Attempts_90D"]
).all()

assert int(priority_mask.sum()) <= int(
    customers["Call_Pressure_90D"]
    .eq("High Relative Pressure").sum()
)
```

```python

# ============================================================
# STEP 5 — FINAL VALIDATION
# ============================================================
n_customers = len(df_customer)
n_outbound = len(outbound_90)
n_observed = int(outbound_90["Outcome_Observed"].sum())
n_answer = int(outbound_90["Recorded_ANSWER"].sum())
n_priority = int(priority_mask.sum())

# Bảo toàn Customer và Outbound Attempts.
assert len(CUSTOMER_FREQUENCY_TABLE) == n_customers
assert CUSTOMER_FREQUENCY_TABLE["SKP_CLIENT"].is_unique

assert (
    CUSTOMER_FREQUENCY_TABLE["Outbound_Call_Attempts_90D"].sum()
    == n_outbound
)

assert (
    CALL_FATIGUE_ANALYSIS["Outbound_Attempts"].sum()
    == n_outbound
)

assert (
    CALL_FATIGUE_ANALYSIS["Observed_Outcomes"].sum()
    == n_observed
)

assert (
    CALL_FATIGUE_ANALYSIS["ANSWER_Count"].sum()
    == n_answer
)

# Kiểm tra coverage quan sát trước đầu cửa sổ 90D.
outbound_source_start = start.loc[outbound_mask].min()
lookback_observed = (
    pd.notna(outbound_source_start)
    and outbound_source_start <= LOOKBACK_START
)

VALIDATION_SUMMARY_22 = pd.DataFrame({
    "Validation Metric": [
        "Snapshot T0",
        "Frequency Window",
        "Prior Outbound Window",
        "Original Customer Rows",
        "Final Customer Rows",
        "JOIN Fan-out",
        "Valid Outbound Attempts (90D)",
        "Observed System Outcomes",
        "ANSWER-coded Calls",
        "Unknown / Unobserved Outcomes",
        "Customers Flagged for Outreach Review",
        "Earlier Outbound History Observed",
        "Positive Response Mapping",
        "Customer / Call Conservation",
        "Future Outcome Exclusion",
    ],
    "Result": [
        str(T0),
        "90 days",
        "7 days before each call",
        f"{n_customers:,}",
        f"{len(CUSTOMER_FREQUENCY_TABLE):,}",
        f"{len(CUSTOMER_FREQUENCY_TABLE) - n_customers:,}",
        f"{n_outbound:,}",
        f"{n_observed:,}",
        f"{n_answer:,}",
        f"{n_outbound - n_observed:,}",
        f"{n_priority:,}",
        "YES" if lookback_observed else "NOT OBSERVED",
        "NOT VERIFIED — RATE NOT CALCULATED",
        "PASS",
        "PASS",
    ],
})

show_bi(
    VALIDATION_SUMMARY_22,
    "FINAL VALIDATION | MULTI-CHANNEL FREQUENCY & CALL FATIGUE",
    "Are customer counts, outbound history and response metrics "
    "consistent at the common snapshot?",
    (
        "Outbound Attempts = valid raw outbound call records; "
        "no CALL_ID deduplication. "
        "Answer Rate = ANSWER-coded / observed system outcomes. "
        "Frequency features from Action 2.1 are reused unchanged. "
        "Annual 2025 App Sessions is descriptive only. "
        "Earlier observed outbound history does not by itself "
        "prove continuous seven-day source coverage. "
        "Positive Response Rate is excluded without verified "
        "RESPONSE_RESULT_2 mapping."
    ),
)

# ============================================================
# UP TO TWO EVIDENCE-BASED BUSINESS INSIGHTS
# ============================================================
insights = []

if observed_decline:
    insights.append({
        "Insight": "Lower ANSWER rate at higher prior call intensity",
        "Evidence": (
            f"Zero-prior: {baseline_rate:.2f}% (n={baseline_n:,}); "
            f"High-prior: {high_rate:.2f}% (n={high_n:,}); "
            f"Delta: {delta_pp:+.2f} pp"
        ),
        "Business Action": (
            "Run a controlled pilot of reduced redial intensity "
            "and alternative calling times."
        ),
    })

high_pressure_n = int(
    customers["Call_Pressure_90D"]
    .eq("High Relative Pressure").sum()
)

if high_pressure_n > 0:
    insights.append({
        "Insight": "High outbound exposure with zero recorded ANSWER",
        "Evidence": (
            f"{n_priority:,} / {high_pressure_n:,} "
            f"high-pressure customers "
            f"({n_priority / high_pressure_n:.2%}); "
            "only complete observed outcomes are flagged"
        ),
        "Business Action": (
            "Prioritize outreach review and test alternative "
            "contact strategies; do not impose an automatic cap."
        ),
    })

INSIGHT_SUMMARY_22 = pd.DataFrame(insights[:2])

if not INSIGHT_SUMMARY_22.empty:
    show_bi(
        INSIGHT_SUMMARY_22,
        "BUSINESS EVIDENCE | OUTREACH OPTIMIZATION",
        "Where is there evidence to review current "
        "outbound contact strategy?",
        (
            "High pressure is defined relative to the observed "
            "customer distribution, not a five-calls-per-week rule. "
            "Associations may reflect customer selection, "
            "call timing, contact history and other confounders. "
            "Recommended actions require controlled evaluation."
        ),
    )
```
